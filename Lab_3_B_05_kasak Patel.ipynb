```python
X = data.iloc[:, :-1].values    # Stores all input features except the last column
y = data.iloc[:, -1].values     # Stores the target/output variable from the last column
```

#### def find_s(X, y): → Defines the Find-S algorithm function.
#### hypothesis = None → Initializes the hypothesis as empty.
#### for i in range(len(X)): → Iterates through all training examples.
#### if y[i].lower() == "yes": → Considers only positive examples.
#### hypothesis = X[i].copy() → Uses the first positive example as the initial hypothesis.
#### hypothesis[j] = "?" → Replaces mismatched attributes with ?.
#### final_hypothesis = find_s(X, y) → Calls the Find-S function.

```python
import numpy as np
def find_s(X, y):
    hypothesis = None
    print("Initial Hypothesis :", hypothesis)
    
    for i in range(len(X)):
        if y[i].lower() == "yes":   # consider only positive examples
            if hypothesis is None:
                hypothesis = X[i].copy()
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != X[i][j]:
                        hypothesis[j] = "?"
        print(f"\n After Training Ex. {i+1}")     
        print(hypothesis)
    return hypothesis

final_hypothesis = find_s(X, y)

print("\n Final Hypothesis")
print(final_hypothesis)
```

### Candidate Elimination maintains the Specific (S) and General (G) boundaries and updates them using positive and negative training examples.
#### def candidate_elimination(concepts, target): → Defines the Candidate Elimination algorithm.
#### specific = concepts[0].copy() → Initializes the specific boundary using the first training example.


```python
def candidate_elimination(concepts, target):
    specific = concepts[0].copy()
    general = [["?" for i in range(len(specific))]]
    print("\ninitial S:", specific)
    print("Initial G:", general)
    
    for i, h in enumerate(concepts):
        if target[i] == "Yes":
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    specific[x] = "?"
        else:
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    general.append(["?" if j != x else specific[x] for j in range(len(specific))])
                    
        print("\n Example", i+1)
        print("S =", specific)
        print("G =", general)
    return specific,general
S, G = candidate_elimination(X,y)
print("\nFinal Specific Boundary")
print(S)
print("\nFinal General Boundary")

for g in G:
    print(g)
```

### Theory-Based Viva Questions & Answers

Q1. What is the main purpose of the Candidate Elimination Algorithm?
Ans: The main purpose is to find all hypotheses that are consistent with the given training data by maintaining the Specific Boundary (S) and General Boundary (G).

Q2. What is the difference between Find-S and Candidate Elimination?
Ans: Find-S considers only positive examples and finds the most specific hypothesis. Candidate Elimination considers both positive and negative examples and maintains both S and G boundaries.

Q3. What does the symbol ? represent in a hypothesis?
Ans: The symbol ? represents a wildcard, meaning that the attribute can have any possible value.

Q4. What are positive and negative examples in concept learning?
Ans: A positive example belongs to the target concept and is usually represented by Yes, while a negative example does not belong to the target concept and is represented by No.

Q5. What are the limitations of the Candidate Elimination Algorithm?
Ans: It can become computationally expensive when the hypothesis space is large. It also requires noise-free and correctly labeled training data to work effectively.
